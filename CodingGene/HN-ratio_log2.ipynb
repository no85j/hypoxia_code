```python
#セットアップと引数の定義
import pandas as pd
import numpy as np
import os

expression_file = './data/TPM.tsv'
pair_file = './data/HN-pair.tsv'
output_prefix = './data/HN-ratio'
```

```python
#入力ファイルの読み込み
df = pd.read_csv(expression_file, sep='\t', index_col=0)
HN = pd.read_csv(pair_file ,sep='\t')
```

```python
#入力ファイルの確認
print("TPM値の確認:")
h = HN.iloc[0, 0]
n = HN.iloc[0, 1]
print(f"{h} の TPM値:")
print(df[h].head(5))
print(f"{n} の TPM値:")
print(df[n].head(5))

print("\nペアファイルの確認:")
print("HN head:")
print(HN.head())
```

```python
#H/N比の計算
results = {}

for i in range(len(HN)):
    h = str(HN.iloc[i,0])
    n = str(HN.iloc[i,1])
    if h in df.columns and n in df.columns:
        colname = f"{h} VS {n}"
        results[colname] = (df.loc[:,h] + 1)/(df.loc[:,n]+1)
    else:
        print(f"警告: {h} または {n} が DataFrameに存在しません")

HNratio = pd.DataFrame(results)
HNratio.insert(0, "GeneName", df.index)
```

```python
#結果の確認
print("HNratioの計算結果(最初の5行):")
print(HNratio.head())
```

```python
#log2変換
HNratio_ = HNratio.set_index('GeneName')
HNratiolog2 = HNratio_.apply(np.log2)
HNratiolog2 = HNratiolog2.reset_index()
```

```python
#結果の確認(log2変換後)
print("HNratiolog2の計算結果(最初の5行):")
print(HNratiolog2.head())
```

```python
#出力
outdir = os.path.dirname(output_prefix)
os.makedirs(outdir, exist_ok=True)
HNratio_.to_csv(f"{output_prefix}.csv", sep=",")
HNratiolog2.set_index('GeneName').to_csv(f"{output_prefix}_log2.csv", sep=",")
print("出力ファイルが正常に保存されました。")
```
