```python
#セットアップと引数の定義
import pandas as pd
import numpy as np
import os

ratio_file = './data/HN-ratio.csv'
thresholds = [1.5, 2.0, 2.5] # 複数指定可能
output_file = './data/HN-score.csv'
```

```python
#入力ファイルの読み込み
df = pd.read_csv(ratio_file, sep=',')
df1 = df.set_index('GeneName')
```

```python
#入力ファイルの確認
print("入力ファイルの確認:")
print(df.head())
```

```python
#HN-scoreの算出
temp = pd.DataFrame()

# 全サンプル数をカウント
temp["all"] = (df1 >= 0).sum(axis=1)

# 各閾値に基づいてHN-scoreを計算
for t in thresholds:
    label = str(t)
    temp[f"up{label}"] = (df1 >= t).sum(axis=1)
    temp[f"down{label}"] = (df1 <= 1/t).sum(axis=1)
    temp[f"unchange{label}"] = temp[f"all"] - temp[f"up{label}"] - temp[f"down{label}"]
    temp[f"HN{label}"] = temp[f"up{label}"] - temp[f"down{label}"]

temp = temp.reset_index()
```

```python
#結果の確認
print("HN-score計算結果(5行):")
print(temp.head())
```

```python
#出力
outdir = os.path.dirname(output_file)
os.makedirs(outdir, exist_ok=True)
temp.to_csv(output_file, sep=",", index=False)
print("出力ファイルが正常に保存されました。")
```
